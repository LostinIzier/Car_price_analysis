```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
import seaborn as sns
```

```python
df_cars=pd.read_csv("cleaned_data.csv")
```

```python
df_cars.columns
```

```python
df_cars["seller"].value_counts()
```

```python
cols=["condition","odometer","mmr","sellingprice","vehicle_age"]
sns.heatmap(df_cars[cols].corr(), annot=True, cmap="coolwarm")
```

#### MMR shows strong correlation with selling price, validating it as a benchmark market valuation.

```python
sns.scatterplot(x="mmr", y="sellingprice", data=df_cars, alpha=0.2)
```

#### Selling price closely follows MMR valuation.

```python
top15 = df_cars["make"].value_counts().head(15).index
df_top15 = df_cars[df_cars["make"].isin(top15)]

sns.boxplot(x="make", y="sellingprice", data=df_top15)
plt.xticks(rotation=45)
plt.sh
```
